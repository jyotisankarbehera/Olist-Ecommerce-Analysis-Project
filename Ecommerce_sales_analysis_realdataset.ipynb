```python
#Average Order Value by Month
#calculate order revenue of each order
order_summary = df.groupby('order_id').agg(
    order_value=('price', 'sum'),
    month=('order_purchase_timestamp', 'first')
)
order_summary['month'] = order_summary['month'].dt.month
#monthly avg order
monthly_aov = order_summary.groupby('month')['order_value'].mean()
monthly_aov
plt.figure(figsize=(12, 6))

month_names = [
    pd.Timestamp(2000, int(m), 1).strftime('%b')
    for m in monthly_aov.index
]

plt.plot(
    month_names,
    monthly_aov.values,
    marker='o',
    linewidth=2.5,
    markersize=8
)

for x, y in zip(month_names, monthly_aov.values):
    plt.annotate(
        f'R$ {y:,.2f}',
        (x, y),
        textcoords='offset points',
        xytext=(0, 10),
        ha='center',
        fontsize=9
    )

plt.title('Monthly Average Order Value (AOV)',
          fontsize=18, fontweight='bold', pad=20)

plt.xlabel('Month', fontsize=12, fontweight='bold')
plt.ylabel('Average Order Value (BRL)', fontsize=12, fontweight='bold')

plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
#Order Delivery Time Distribution
delivery_data = df[
    df['order_status'] == 'delivered'
][[
    'order_id',
    'order_purchase_timestamp',
    'order_delivered_customer_date'
]].drop_duplicates(subset='order_id')

delivery_data['delivery_days'] = (
    delivery_data['order_delivered_customer_date'] -
    delivery_data['order_purchase_timestamp']
).dt.days

delivery_data = delivery_data.dropna(subset=['delivery_days'])

delivery_data['delivery_days'].describe()


plt.figure(figsize=(10, 5))

plt.hist(
    delivery_data['delivery_days'],
    bins=20,
    color='darkorange',
    edgecolor='white'
)

plt.title('Order Delivery Time Distribution')
plt.xlabel('Delivery Time (Days)')
plt.ylabel('Number of Orders')

plt.grid(axis='y', alpha=0.25)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
#Actual vs Estimated Delivery Time
delivery_compare = df[df['order_status'] == 'delivered'][
    ['order_id', 'order_purchase_timestamp',
     'order_delivered_customer_date',
     'order_estimated_delivery_date']
].drop_duplicates(subset='order_id')

delivery_compare['actual_days'] = (
    delivery_compare['order_delivered_customer_date'] -
    delivery_compare['order_purchase_timestamp']
).dt.days

delivery_compare['estimated_days'] = (
    delivery_compare['order_estimated_delivery_date'] -
    delivery_compare['order_purchase_timestamp']
).dt.days

delivery_compare[['actual_days', 'estimated_days']].mean()

avg_delivery = [
    delivery_compare['actual_days'].mean(),
    delivery_compare['estimated_days'].mean()
]

labels = ['Actual Delivery', 'Estimated Delivery']

plt.figure(figsize=(8, 5))

plt.bar(
    labels,
    avg_delivery,
    color=['steelblue', 'darkorange'],
    width=0.5
)

plt.title('Actual vs Estimated Delivery Time')
plt.xlabel('Delivery Type')
plt.ylabel('Average Days')

plt.grid(axis='y', alpha=0.25)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
#On-Time vs Delayed Deliveries
delivery_status = df[df['order_status'] == 'delivered'][
    ['order_id', 'order_delivered_customer_date',
     'order_estimated_delivery_date']
].drop_duplicates(subset='order_id')

delivery_status = delivery_status.dropna()

delivery_status['delivery_result'] = delivery_status.apply(
    lambda row: 'On-Time' if row['order_delivered_customer_date'] <= row['order_estimated_delivery_date'] else 'Delayed',
    axis=1
)

delivery_count = delivery_status['delivery_result'].value_counts()
delivery_count

plt.figure(figsize=(8, 5))

plt.bar(
    delivery_count.index,
    delivery_count.values,
    color='mediumpurple',
    width=0.55
)

plt.title('On-Time vs Delayed Deliveries')
plt.xlabel('Delivery Status')
plt.ylabel('Number of Orders')

plt.grid(axis='y', alpha=0.25)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

# Business Insights

## 1. Customer Payment Preference
Credit Card was the most commonly used payment method, followed by Boleto, Voucher, and Debit Card. This indicates that customers preferred credit card payments over other available payment methods.

## 2. Product Category Performance
Bed, Bath & Table had the highest number of orders among the product categories, followed by Health & Beauty and Sports & Leisure. This indicates strong customer demand for these product categories.

## 3. Order Status Analysis
The dataset contains a very high number of delivered records compared with canceled records. This indicates that most recorded orders were successfully delivered.

## 4. Purchase Day Analysis

Monday recorded the highest number of purchase records, with 18,458 records. This indicates that customer purchasing activity was relatively high at the beginning of the week.

## 5. State-wise Customer Distribution

São Paulo (SP) had the highest number of unique customers, with 38,641 customers, followed by Rio de Janeiro (RJ) with 11,746 and Minas Gerais (MG) with 10,846. This indicates that customer concentration was highest in São Paulo.

## 6. City-wise Customer Distribution

São Paulo had the highest number of unique customers, with 14,310 customers, followed by Rio de Janeiro with 6,263 and Belo Horizonte with 2,562. This indicates that São Paulo was the leading customer market among the cities analyzed.

## 7. Top Products by Revenue

The product with ID `bb50f2e236e5eea0100680137654686c` generated the highest sales revenue of R$ 68,160.00, followed by product IDs `6cdd53843498f92890544667809f1595` and `d6160fb7873f184099d9bc95e30376af`. This indicates that a small group of products contributed significantly to the total sales revenue.

## 8. Monthly Average Order Value (AOV)

September recorded the highest average order value at approximately R$ 155.79, while February had the lowest at approximately R$ 134.76. This indicates that customers spent more per order on average in September compared with February.

## 9. Average Order Delivery Time

The average order delivery time was approximately 12.09 days. This indicates that customers received their orders in around 12 days on average after purchase.

## 10. On-Time vs Delayed Deliveries

The analysis shows that 87,417 orders were delivered on time, while 7,705 orders were delayed. This indicates that most delivered orders reached customers on or before the estimated delivery date, showing good delivery performance.

## 11. Top Sellers by Revenue

Seller ID `53243585a1d6dc2643021fd1853d8905` generated the highest sales revenue of BRL 239,791.94, followed by seller ID `4869f7a5dfa277a7dca6462dcf3b52b2` with BRL 235,382.53. This indicates that a small group of sellers contributed significantly to the platform's sales revenue.

# Project Limitations

1. **Missing Product Cost Data:** The dataset does not contain product cost information. Therefore, actual profit and loss could not be calculated.

2. **Limited Product Information:** Product names are not available for all products, so Product IDs were used in some analyses.

3. **Limited Seller Information:** Seller names were not available, so Seller IDs were used to identify top-performing sellers.

4. **Historical Data:** The analysis is based on historical e-commerce data and may not represent current customer behavior or sales trends.

5. **Dataset Structure:** The dataset contains multiple rows for some orders because each order may include multiple products. Therefore, suitable grouping and deduplication methods were used for order-level analysis.
